<h2> Introduction </h2>

This module aims to give an introduction to the art of computer programming.  <b>Computer programming</b> refers to the construction of sets of instructions that tell a computer which sequence of operations it should perform, and on which data it should perform them.  This skill is indispensable for mathematicians, as so many calculations are beyond the scope of the human brain, and certainly beyond the limit of human patience!

Computer programming, or <em>coding</em>, as we shall sometimes refer to it, allows us to compose scripts that implement  sophisticated numerical routines.  These allow us to do things as diverse as solve combinatorial problems by combing through arrangements of objects, approximate the solutions to differential equations, configure and analyse large sets of data, and automate algorithms to solve linear systems.

Here however, we start with the fundamentals.  Coding can only be practised through the use of some <b>programming language</b>.  A programming language is a system of notation that assigns a set of human-readable instructions to common logical and arithmetic operations.  The language can exist entirely independently of a computer.

In practice however, we normally engage with a programming language via a piece of software called an <b>interpreter</b>.  An interpreter translates the human-readable instructions into machine code, which is then given to a central processing unit capable of reading this code and executing the arithmetic and logical operations contained within.  The results are then returned to us via the software.

```python
print(1)
print(2)
print(3)
# ...

# They can also be negative.

print(-1)

# They can be astonishingly large.

# print(2**10000)

# (The above demonstrates another common use of comments,
# to stop unwanted code from being run!)
```

<h2>Python as a Calculator</h2>

Python is a computer programming language, and would therefore be of extremely limited use if we could not perform basic computations. Thankfully, Python can handle all basic arithmetic operations.

All arithmetic operators in the Python environment are highlighted in an elegant <span style="color:rgb(170,34,255);font-weight:bold;">purple</span>.

```python
# We have already seen addition.  The operation is performed with the '+' character.
print(1 + 1)

# We have already seen multiplication.  The operation is performed with the '*' character.
print(2*3)

# We can also subtract.  This operation is performed with the '-' character.
print(7 - 2)
```

Note that we didn't include division in the above code cell.  This is because Python allows for many different interpretations of the concept of 'division'.  The first we will mention is <b>float division</b>, so called as the result is always a <samp>float</samp>.

It can be performed by placing the <span style="font-family:consolas;color:rgb(170,34,255);font-weight:bold;">/</span> character between the dividend and the divisor.

```python
print(100/7)
```

This is perhaps the most obvious interpretation of 'division'.  It returns a value that is equal to the ratio of the two values, accurate to as many decimal places as the interpreter will allow.

Note that when we say the result is 'always a float', it is <em>always</em> a float.

```python
print(7/7)
type(7/7)
```

Even when we know the result to be an integer (in the mathematical sense, at least), the interpreter insists that it conforms to the expected data type, a float that just happens to have a fractional part of zero.

Next, there is <b>integer division</b> (sometimes called <b>floor division</b>).  This is similar to floor division, only the value of the ratio is returned to the nearest integer.

It is performed by placing two slashes, <span style="font-family:consolas;color:rgb(170,34,255);font-weight:bold;">//</span> between the dividend and the divisor.

```python
print(100//7)

# 100 = 14*7 + 2
```

Finally, we have an operation that complements floor division, the <b>modulo</b> operation.  This operation returns the remainder obtained after one value is divided by another.

It is performed by placing the <span style="font-family:consolas;color:rgb(170,34,255);font-weight:bold;">%</span> character between the dividend and the divisor.

```python
print(100%7)

# 100 = 14*7 + 2
```

This concludes division.  Our last arithmetic operation is <b>exponentiation</b>, performed by placing two asterisks <span style="font-family:consolas;color:rgb(170,34,255);font-weight:bold;">**</span> between the base value and the power to which it is being raised.

```python
print(2**10)
```

Be cautious here, many programming languages use the character <span style="font-family:consolas;color:rgb(170,34,255);font-weight:bold;">^</span> to denote exponentiation, but in Python, this is an entirely different operation.

```python
print(2^10)
```

(It's something called a 'bitwise operator', but is not something that we'll need for this course, so we'll forgo any more detail.)

The remaining details are pretty straightforward.  As with function calls, it's possible to embed the result of an arithmetic operation as an argument of another arithmetic operation.

```python
print(3 + 4*2)
```

Just remember that, as always, we follow the rules of <a href="https://en.wikipedia.org/wiki/Order_of_operations">BODMAS/BIDMAS/PEMDAS</a>.

```python
print(3 + 4*2)

# The product is computed before the sum.

print((3 + 4)*2)

# The sum is computed before the product.  This requires bracketing.
```

<h2>Variables</h2>

We have now seen how to handle numerical computations in Python.  The question remains; how do we actually <em>use</em> the results of these calculations?

For instance, suppose I want to compute the ratio required to convert miles to kilometres.  If we are told that 28 miles is approximately 45 kilometres, then we can evaluate the conversion factor between these measurements.

```python
45/28
```

Hence, in order to change a length measured in miles to one measured in km, it suffices to multiply the measurement by a factor of 1.607....

But do we really need to recompute this ratio <em>every time</em> we wish to use it?  This could make our code look cluttered and difficult to read.  Instead, we can assign this value to a <b>variable</b>.

Variables are objects that allow us to store data in the system's local memory so that it can be used in later computations.  To assign a piece of data to a variable, we simply decide on a name for that variable, and place an equal sign <span style="font-family:consolas;color:rgb(170,34,255);font-weight:bold;">=</span> between the variable and the data being assigned to it.

```python
miles_to_km = 45/28

# We have included spacing between the variable miles_to_km and the value 45/28.
# This is entirely optional, and is ignored by the interpreter.
```

Running the above cells binds the value returned by <code><span style="color:rgb(0,136,0);">45</span><span style="color:rgb(170,34,255);font-weight:bold;">/</span><span style="color:rgb(0,136,0);">28</span></code> to the variable <samp>miles_to_km</samp>.  Such a command is called an <b>assignment</b>.  Henceforth, whenever this variable is called, the assigned value is returned.

```python
miles_to_km
```

We note something interesting about the above cells.  The first cell contained the computation <code><span style="color:rgb(0,136,0);">45</span><span style="color:rgb(170,34,255);font-weight:bold;">/</span><span style="color:rgb(0,136,0);">28</span></code>, and a <samp>float</samp> was returned.  The second cell however, contained the assignment <code>miles_to_km <span style="color:rgb(170,34,255);font-weight:bold;">=</span> <span style="color:rgb(0,136,0);">45</span><span style="color:rgb(170,34,255);font-weight:bold;">/</span><span style="color:rgb(0,136,0);">28</span></code>, and nothing was returned.  This shows the difference between two fundamental elements of Python's syntax:

<ul>
  <li> Any code that can be evaluated to produce some piece of output data is called an <b>expression</b>. </li>
  <li> Any code that performs an instruction, but does not return an output, is called a <b>statement</b>. </li>
</ul>

Returning to the examples above, the ratio <code><span style="color:rgb(0,136,0);">45</span><span style="color:rgb(170,34,255);font-weight:bold;">/</span><span style="color:rgb(0,136,0);">28</span></code> is an expression, whereas the assignment <code>miles_to_km <span style="color:rgb(170,34,255);font-weight:bold;">=</span> <span style="color:rgb(0,136,0);">45</span><span style="color:rgb(170,34,255);font-weight:bold;">/</span><span style="color:rgb(0,136,0);">28</span></code> is a statement.

Returning to the practical use of variables, they can not only be called, but they can be called within larger expressions.  Suppose we wanted to evaluate how many kilometres were in 5 miles, we would need to multiply our conversion factor <samp>miles_to_km</samp> by <samp>5</samp>.

```python
5*miles_to_km
```

Expressions can even be evaluated in the same code cell in which the variables were defined.

```python
a = 37   # Only the result of the final expression will be returned.
b = 45   # But this is fine, as nothing is returned from these statements anyway.

a + 3*b  # The above assignments were still run however, so this expression makes sense.
```

As is always the case with coding, there are a few subtleties.  It's here that we note the important point that unless a code cell is actually run, the code inside will not be read by the interpreter. If this seems complicated, it gets stranger.  The interpreter only computes the instructions given in the cells being run.  It also performs them in the exact order in which we run them.  This means that it's possible to define a variable in a cell at the bottom of a notebook, run that cell, and then go back and use that variable in a code cell at the top of the notebook.

However, just because something is <em>possible</em>, does not necessarily mean it is <em>desirable</em>.  No responsible programmer should ever do this unless it is absolutely necessary.  Indeed, most notebooks will be run from top to bottom.  There is even a function in the <kbd>Kernel</kbd> menu that runs notebooks from top to bottom automatically.

Moving on from these complications, it is possible to define variables using computations that use <em>other variables</em>.

```python
km_to_miles = 1/miles_to_km
```

```python
# Test this by multiplying these two inverse conversion factors together to get one.

km_to_miles*miles_to_km
```

If we need to redefine a variable to a new value, then it's possible to overwrite a previously defined variable by using the same syntax as the original assignment.

```python
miles_to_km = 1.60934

# This is actually more accurate than 45/28 was.
```

```python
miles_to_km
```

But be careful here, if the original code cell that defined <samp>miles_to_km</samp> had been deleted, then we would have no way of retrieving the orignal value of <samp>miles_to_km</samp> before it was redefined.  Be mindful of overwriting data, is it really necessary?

Another note, even though <samp>km_to_miles</samp> was defined as <code><span style="color:rgb(0,136,0);">1</span><span style="color:rgb(170,34,255);font-weight:bold;">/</span>miles_to_km</code>, the value of <samp>km_to_miles</samp> has <em>not</em> been updated to reflect the new value of <samp>miles_to_km</samp>.

```python
km_to_miles*miles_to_km

# If the km_to_miles value had also been updated, then this product should be one.
# But it isn't!
```

```python
km_to_miles = 1/miles_to_km

# However, nothing stops us from redefining km_to_miles to reflect this change.

km_to_miles*miles_to_km
```

On the subject of overwriting variables, it is possible for variables to overwrite <em>themselves</em>.

```python
print(a)

a = a + 1

print(a)
```

This would make sense in situations where we needed a variable to act as a counter.  Such commands are so frequently needed that there are special types of assignment syntaxes dedicated to them, the <b>augmented assignments</b>.  To reassign a variable as its current value plus some new value, we use the combination <span style="font-family:consolas;color:rgb(170,34,255);font-weight:bold;">+=</span>.

```python
a += 1

print(a)
```

Similarly, we have <span style="font-family:consolas;color:rgb(170,34,255);font-weight:bold;">-=</span>, <span style="font-family:consolas;color:rgb(170,34,255);font-weight:bold;">\*=</span>, <span style="font-family:consolas;color:rgb(170,34,255);font-weight:bold;">/=</span>, <span style="font-family:consolas;color:rgb(170,34,255);font-weight:bold;">//=</span>, <span style="font-family:consolas;color:rgb(170,34,255);font-weight:bold;">%=</span> and <span style="font-family:consolas;color:rgb(170,34,255);font-weight:bold;">**=</span>.

```python
a **= 2  # a is reassigned to the square of its current value.

print(a)
```

We clearly see our <samp>miles_to_km</samp> and <samp>km_to_miles</samp> variables at the bottom of the list.  There are also a lot of other variables that we didn't define.  These are pre-defined variables that Python and Jupyter generate so that built-in functions and operations work correctly.

For example, the <samp>In</samp> variable contains the contents of every code cell that we have run so far in this coding session.  It is defined automatically by the Jupyter Notebook software at the start of each session.

Any variable name that starts with an underscore is a private variable that is fundamental to the operation of the interactive Python environment.  For that reason, we recommend not using variable names that start with this character.

    <li> Names that are misleading (intentionally or otherwise).</li>

This last case is particularly troublesome, as if we assign a value to the name of a built-in function, we will lose the ability to call that function.

```python
print(3.142)
```

This last call has resulted in a <span style="font-family:consolas;color:rgb(178,43,49);font-weight:bold;">TypeError</span> because we were attempting to use <span style="font-family:consolas;color:rgb(0,136,0);">print</span> as a function, but we just reassigned the variable to the string <span style="font-family:consolas;color:rgb(186,33,33);">'hello'</span>.

```python
print
```

Thankfully, there is a quick and convenient way of retrieving the original function if this happens.  We can use the <span style="font-family:consolas;color:rgb(0,128,0);font-weight:bold;">del</span> keyword that we saw earlier.  Normally, this would delete the variable entirely, but if the variable is normally assigned to a built-in function, then the <span style="font-family:consolas;color:rgb(0,128,0);font-weight:bold;">del</span> keyword restores this assignment.

```python
del print

print(3.142)
```
